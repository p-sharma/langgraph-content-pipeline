```python
from langchain_tavily import TavilySearch
```

```python
tool = TavilySearch(max_results = 5, topic="general")
raw = tool.invoke({"query": "Rust in the Linux kernel"})

print(raw.keys())
raw["results"][0]
```

```python
from content_agent.tools.search import search_web, search_many

results = search_web("Rust in the Linux kernel")
for r in results:
    print(r["title"], "→", r["url"])
```

```python
queries = [
    "Rust in the Linux kernel",
    "Rust Linux kernel drivers progress",
    "Rust Linux kernel maintainers debate",
]
combined = search_many(queries)
print(f"{len(combined)} unique results from {len(queries)} queries")
```

```python
import json

with open("research/sample_search.json", "w") as f:
    json.dump(combined, f, indent=2)
```
